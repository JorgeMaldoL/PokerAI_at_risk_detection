# Bustabit Data Exploration

Bustabit is an online gambling website that collects data on its actual player base. 

### Gameplay

    Players make bets. There is a multiplier that starts at 1.00x and increases (1.10, 1.50, 2.00, 3.00, etc). The player can click "cash out" to lock in the desired multiplier. The winnings from your bet are multiplied by the same multiplier the player "cashed out" at. 

    There is a bust mechanic to the game, where in any moment the game is being played their is a chance the multiplier stops increasing and does a bust to 0. All players that has not cashed out before that happen lose their bet. 

    The longer the player waits to cash out, the higher the multiplier gets, but also the higher the risk of losing. 

#### The following is a initial look at the raw dataset: structure, summary statistics, and a sample of rows.

```python
import pandas as pd

from poker_coach.utils import rainbow
from poker_coach.config import ID_COLS, BUSTABIT_DATA_FILE
```

```python
df = pd.read_csv(BUSTABIT_DATA_FILE, parse_dates=['PlayDate'])
df['Won'] = df['CashedOut'].notna()   # missing CashedOut = player busted (lost)
print(f"{len(df)} bets from {df['Username'].nunique()} players")
rainbow(df.drop(columns=ID_COLS).describe())
```

### What each column represents

    Id: This is the unique ID for each bet transaction that happened on the website. 

    GameID: That game rounds unique ID. When multiple  players bet on the same round, you'll  see the same Game ID. 

    Username: The username associated with that player's account. 

    Bet: This is the amount of money the player wagered during that playthrough. 

    CashedOut: This is the multiplier that the player cashed out on, effectively exiting the playthrough (Total winnings = money wagered x multiplier). If it says NaN, then they lost. 

    Bonus: Any extra money that was added to the winnings, along with the cashout. NaN just  means they lost.

    Profit: How much the player profited from that bet (total winnings - money wagered = profit). If  the profit says NaN, it means there's nothing to calculate.

    BustedAt: The multiplier that the game crashed that round. (All players who haven't cashed out before this number lost their bet)

    PlayDate: This is a time and date when the bet happened
